## Hashing for Student Codes

```python
#%pip install hashlib

import hashlib
```

```python
# Open the file in read mode and read its lines into a list
with open('/Users/ramius/Desktop/CodeVault/01_Project/Work/Susumu/Student_Survey/StudentSurvey_WS2526/_rooms/code_list.txt', 'r') as file:
    lines = file.readlines()

# Optionally, you can remove the newline characters from each line
codes = [line.strip() for line in lines]
#codes.append("x")
```

```python
assigned_codes = set()  # only stores used codes, not student IDs
assigned_ids = set()

def get_unique_code(student_id):
    for attempt in range(len(codes)):
        salted = f"{student_id}-{attempt}"
        hash_digest = hashlib.sha256(salted.encode()).hexdigest()
        index = int(hash_digest, 16) % len(codes)
        code = codes[index]
        if code not in assigned_codes and student_id not in assigned_ids:
            assigned_codes.add(code)
            assigned_ids.add(student_id)
            return code
    raise Exception("No unique codes left for your ID!")
```

```python
# Example
code = get_unique_code("12345677")
print(code)
```

```python
code = get_unique_code("1406810")
print(code)
```
